# Job Candidates

Suppose a firm has identified 16 candidates to fill 4 positions open at the company. The candidates were evaluated equally and without regard for gender. Of those 16 candidates, 10 are male and 6 are female. If the positions were filled at random, would it be unusual if all of them were male?

To determine if filling the positions randomly yields all males is considered *unusual*, we need to compute the probability of all males filling the positions. To compute this probability, we need to cound the number of was the candidates can fill the positions, and the number of ways the males can fill the positions.

The total number ways 16 candidates can fill 4 positions is
$$
\begin{pmatrix} 16 \\ 4 \end{pmatrix} = \frac{16!}{4!(16-4)!} = 1820
$$

The total number of ways 10 male candidates can fill 4 positions is
$$
\begin{pmatrix} 10 \\ 4 \end{pmatrix} = \frac{10!}{4!(10-4)!} = 210
$$

Hence the probability of all males is
$$
P(\text{all males}) = \frac{210}{1820} = 0.1154
$$

Because the probability for all males is $11.54\%$, it would be unusual if all males filled the positions.

```python
import numpy as np

candidates = [
    'M', 'M', 'M', 'M', 'M', 'M', 'M', 'M', 'M', 'M',
    'F', 'F', 'F', 'F', 'F', 'F',
]

B = 15_000
count = 0

for b in range(B):
    positions = np.random.choice(candidates, size=4, replace=False)
    count += 1 if 'F' not in positions else 0

p_males = count / B
print(f"P(all males) = {p_males:.2%}")
```
